```python
import pandas as pd

columnas = [
    "seqname",
    "source",
    "feature",
    "start",
    "end",
    "score",
    "strand",
    "phase",
    "attributes"
]

try:
    gff = pd.read_csv(
        "../datos/human_genes.gff",
        sep="\t",
        comment="#",
        header=None,
        names=columnas
    )

except FileNotFoundError:
    print("No se encontró el archivo human_genes.gff")
    
tmpdf = gff["attributes"]

def parse_attributes(attributes):
    info = {}

    fields = attributes.split(";")

    for field in fields:
        if "=" in field:
            key, value = field.split("=",1)
            info[key] = value

    return info

gene_dict={}
for attributes in gff["attributes"]:
    info = parse_attributes(attributes)

    gene_name = info.get("Name")
    description = info.get("description")

    if gene_name:
        gene_dict[gene_name] = description
    
```
